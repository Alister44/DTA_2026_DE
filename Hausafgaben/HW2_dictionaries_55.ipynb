# Kapitel 1. Wörterbücher (`dict`)

**Kurze Wiederholung:**

Ein Wörterbuch ist eine Sammlung von `Schlüssel : Wert`-Paaren, erstellt in geschweiften Klammern:
```python
student = {"name": "Olena", "age": 20, "city": "Berlin"}
```

**Grundlegende Operationen:**
- Zugriff: `student["name"]`
- Hinzufügen/Ändern: `student["email"] = "olena@example.com"`
- Löschen: `del student["age"]`
- Prüfung: `if "name" in student:`
- Sicherer Zugriff: `student.get("phone")` — gibt `None` zurück, wenn der Schlüssel nicht existiert
- Länge: `len(student)` — Anzahl der Paare

**Durchlaufen in einer Schleife:**
- `for key in d.keys():` — Durchlaufen der Schlüssel
- `for value in d.values():` — Durchlaufen der Werte
- `for key, value in d.items():` — Durchlaufen der Paare

````python
text = "Katze Hund Katze Hamster Katze Hund Papagei Katze Hund Fisch"
# Dein Code:
words = text.split()
word_count = {}

for word in words:
    if word not in word_count.keys():
        word_count[word] = 1
    else:
        word_count[word] = word_count[word] + 1

print(word_count)

````

### Aufgabe 22. Telefonbuch mit Suche ⭐⭐⭐

Erstelle ein Telefonbuch-Wörterbuch mit 5 Kontakten. Schreibe ein Programm, das in einer `while`-Schleife:
1. Den Benutzer nach einer Aktion fragt: `suchen`, `hinzufügen`, `löschen`, `beenden`
2. Je nach Auswahl die entsprechende Aktion ausführt
3. Endet, wenn der Benutzer `beenden` eingibt

> 💡 **Tipp:** Verwende `while True:` (Endlosschleife) und eine Abbruchbedingung mit `break`, ODER eine Flag-Variable, z. B. `running = True`.


````python
# Dein Code für Aufgabe 22:
phone_book = {
    "Anja":   "+491991234567",
    "Borja":  "+491932345678",
    "Wasja":  "+491503456789",
    "Dascha":  "+491674567890",
    "Shenja":  "+491445678901"
}

while True:
    act = input("Aktion eingeben (suchen, hinzufügen, löschen, beenden): ")
    if act == 'suchen':
        name = input("Namen eingeben: ")
        if name in phone_book:
            print(name, phone_book[name])
        else:
            print("Name nicht gefunden!")
    elif act == 'hinzufügen':
        name_new = input("Namen eingeben: ")
        if name_new not in phone_book:
            number_new = input("Telefonnummer eingeben: ")
            phone_book[name_new] = number_new
        else:
            print("Dieser Name existiert bereits")
    elif act == 'löschen':
        name_del = input("Namen des zu löschenden Kontakts eingeben: ")
        if name_del in phone_book:
           del phone_book[name_del]
           print("Löschen erfolgreich")
        else:
            print("Diesen Kontakt gibt es nicht")
    elif act == "beenden":
        break
    else:
        print("Aktion nicht gefunden!")



````

### Aufgabe 23. Wörterbuch nach Wortlänge ⭐⭐⭐

Gegeben ist eine Liste von Wörtern. Erstelle ein Wörterbuch, bei dem der Schlüssel die Wortlänge ist und der Wert eine Liste der Wörter dieser Länge.

**Beispiel:** für `["Hund", "Katze", "Löwe", "Fuchs"]` → `{4: ["Hund", "Löwe"], 5: ["Katze", "Fuchs"]}`

> 💡 **Tipp:** Prüfe, ob der Schlüssel-Länge im Wörterbuch schon existiert. Wenn ja — füge das Wort mit `.append()` zur bestehenden Liste hinzu. Wenn nicht — erstelle einen neuen Schlüssel mit einer Liste, die ein Wort enthält.


````python
words = ["Katze", "Programm", "Sonne", "Hund", "Computer", "Tisch", "Löwe", "Nacht", "Schule", "Tag"]
# Dein Code:
dict_words = {}
for word in words:
    if len(word) not in dict_words:
        dict_words[len(word)] = [word]
    else:
        dict_words[len(word)].append(word)
dict_words
````

### Aufgabe 25. Anwesenheitserfassung ⭐⭐⭐

Gegeben ist ein Wörterbuch mit der Anwesenheit von Studenten für eine Woche: Schlüssel — Name, Wert — eine Liste mit 5 Werten (nach Tagen, von Montag bis Freitag). 1 bedeutet anwesend, 0 — abwesend.

Gib aus:
1. Wie viele Tage jeder Student anwesend war
2. Wer alle 5 Tage anwesend war
3. An welchem Tag die meisten Fehltage waren
4. Die gesamte Anwesenheitsquote der Gruppe (Summe der 1en / Gesamtzahl der Werte × 100)

> 💡 **Tipp:** Für Punkt 3: erstelle eine Hilfsliste `absences = [0, 0, 0, 0, 0]` (nach Tagen), durchlaufe die Werte des Wörterbuchs und addiere dazu. Finde dann den Index des Maximums.


````python
attendance = {
    "Olena":   [1, 1, 1, 1, 1],
    "Andrij":  [1, 0, 1, 1, 0],
    "Maria":   [1, 1, 1, 1, 1],
    "Bohdan":  [0, 1, 0, 1, 1],
    "Sofia":   [1, 1, 1, 0, 1],
    "Petro":   [0, 0, 1, 1, 1],
    "Darja":   [1, 1, 1, 1, 0],
    "Mychajlo": [1, 0, 0, 1, 1]
}
# Dein Code:
print("Wie viele Tage jeder Student anwesend war")
for name, info in attendance.items():
    print(name,sum(info))

print("Wer alle 5 Tage anwesend war")
for name, info in attendance.items():
    if sum(info) == 5:
        print(name)


absences = [0, 0, 0, 0, 0]
for info in attendance.values():
    for day in range(0,5):
        if info[day] == 0:
            absences[day] += 1
print(absences)
print(f"{absences.index(max(absences))+1}. Tag - die meisten Fehltage")
````

### Aufgabe 32. Gruppenstatistik ⭐⭐

Gegeben ist ein Wörterbuch mit Studenten und ihren Noten. Berechne und gib aus:

````python
grades = {
    "Olena":   [85, 90, 78, 92, 88],
    "Andrij":  [72, 65, 80, 75, 70],
    "Maria":   [95, 92, 88, 90, 94],
    "Bohdan":  [60, 65, 70, 55, 68],
    "Sofia":   [82, 85, 80, 88, 84],
    "Petro":   [78, 75, 82, 80, 76],
    "Darja":   [90, 88, 92, 85, 91]
}
# Dein Code:
d_avg_student = {}
print("Average score of each student: ")
for name, info in grades.items():
    avg_student = sum(info) / len(info)
    d_avg_student[name] = avg_student
    print(f"{name}, {avg_student}")

avg_group = sum(d_avg_student.values()) / len(d_avg_student)
print("Overall group average score ", avg_group)

print("Number of students with an average score above the general average ")

count = 0
for avg in d_avg_student.values():
    if avg > avg_group:
        count += 1
print(count)

print("The student with the highest GPA: ")
max(d_avg_student, key = lambda x: d_avg_student[x])


````

### Aufgabe 54. Geschäft mit Filtern ⭐⭐⭐

Gegeben ist eine Produktliste — Wörterbücher mit den Feldern `name`, `category`, `price`, `rating`, `in_stock`. Frage in einer `while`-Schleife den Benutzer nach einem Filter:
- `kategorie` — Kategoriename erfragen, Produkte dieser Kategorie ausgeben
- `preis` — Höchstpreis erfragen, günstigere Produkte ausgeben
- `bewertung` — Produkte mit einer Bewertung ≥ 4.5 ausgeben
- `verfügbar` — Produkte ausgeben, bei denen `in_stock = True`
- `top` — die 3 teuersten Produkte ausgeben
- `beenden` — beenden


````python
products = [
    {"name": "Laptop HP",     "category": "Elektronik", "price": 22000, "rating": 4.6, "in_stock": True},
    {"name": "Laptop Lenovo", "category": "Elektronik", "price": 18500, "rating": 4.3, "in_stock": True},
    {"name": "Telefon Samsung","category": "Elektronik", "price": 12000, "rating": 4.7, "in_stock": False},
    {"name": "Buch Python",   "category": "Bücher",       "price": 450,   "rating": 4.8, "in_stock": True},
    {"name": "Buch SQL",      "category": "Bücher",       "price": 380,   "rating": 4.2, "in_stock": True},
    {"name": "Sneaker Nike",  "category": "Kleidung",        "price": 2800,  "rating": 4.5, "in_stock": True},
    {"name": "Jacke Adidas",  "category": "Kleidung",        "price": 3500,  "rating": 4.0, "in_stock": False},
    {"name": "Kopfhörer",      "category": "Elektronik", "price": 1200,  "rating": 4.6, "in_stock": True},
    {"name": "Wasserkocher",         "category": "Haushaltsgeräte",    "price": 850,   "rating": 4.4, "in_stock": True},
    {"name": "Mixer",        "category": "Haushaltsgeräte",    "price": 1500,  "rating": 4.7, "in_stock": False}
]
print("""
Filter:
kategorie — Produkte dieser Kategorie
preis — Höchstpreis eingeben (Ausgabe - Produkte nicht teurer als dieser Preis)
bewertung — Ausgabe von Produkten mit einer Bewertung ≥ 4.5
verfügbar — Produkte, bei denen in_stock = True
top — 3 teuerste Produkte
beenden — beenden\n""")

filter = ''
while filter != 'beenden':
    filter = input("Filternamen eingeben: ")
    if filter == 'kategorie':
        category = input("Kategoriename eingeben: ")
        for product in products:
            if product['category'] == category:
                print(product['name'])
    elif filter == 'preis':
        max_price = int(input("Höchstpreis eingeben: "))
        for product in products:
            if product['price'] < max_price:
                print(product['name'],product['price'])
    elif filter == 'bewertung':
        for product in products:
            if product['rating'] >= 4.5:
                print(product['name'])
    elif filter == 'verfügbar':
        for product in products:
            if product['in_stock'] == True:
                print(product['name'])
    elif filter == 'top':
        sorted_products = sorted(products, key=lambda s: s["price"],reverse = True)
        for product in sorted_products[0:3]:
            print(product['name'],product['price'])



````
