````python
# Dein Code für Aufgabe 27:
num = 50
for i in range(2, num + 1):
  is_prime = True
  for j in range (2, i):
    if i % j == 0:
      is_prime = False
      break
  if is_prime == True:
    print(i)




````

### Aufgabe 42. Summe der eingegebenen Zahlen ⭐⭐

Bitte den Benutzer, Zahlen einzugeben, bis er 0 eingibt.
Gib danach die Summe aller eingegebenen Zahlen aus (Null wird nicht mitgezählt).

### Aufgabe 46. Größter gemeinsamer Teiler ⭐⭐⭐

Frage den Benutzer nach zwei positiven Zahlen und finde ihren größten gemeinsamen Teiler (ggT) mit dem euklidischen Algorithmus.

**Algorithmus:** solange `b ≠ 0`: gleichzeitig `a, b = b, a % b`. Wenn `b = 0`, ist die Antwort `a`.

**Beispiel:** ggT(48, 36) = 12.


````python
# Dein Code für Aufgabe 46:
a = int(input("Gib die erste positive Zahl ein: "))
b = int(input("Gib die zweite positive Zahl ein: "))

while b != 0:
    a, b = b, a % b

print("ggT:", a)
````

### Aufgabe 51. Sortieren und Umkehren ⭐⭐

````python
numbers = [42, 17, 89, 3, 56, 91, 24, 8, 73]
# Dein Code:
n = len(numbers)

for i in range(n):
    for j in range(n - 1):
        if numbers[j] > numbers[j + 1]:
            temp = numbers[j]
            numbers[j] = numbers[j + 1]
            numbers[j + 1] = temp

print("Aufsteigend:", numbers)
reversed_list = []
for i in range(len(numbers) - 1, -1, -1):
    reversed_list.append(numbers[i])
print("In umgekehrter Reihenfolge:", reversed_list)
````

### Aufgabe 55. Listen zusammenführen ⭐⭐⭐

Gegeben sind zwei Listen. Erstelle eine neue Liste, die enthält:
1. Zuerst die Elemente der ersten
2. Dann — die Elemente der zweiten
3. Danach — nur die Elemente aus beiden Listen, die in beiden vorkommen

**Beispiel:** `[1, 2, 3]` + `[2, 3, 4]` → `[1, 2, 3, 2, 3, 4, 2, 3]` (weil 2 und 3 gemeinsam sind).


````python
list_a = [1, 2, 3, 4, 5]
list_b = [3, 4, 5, 6, 7]
# Dein Code:
result = list_a.copy()
for item in list_b:
    result.append(item)
common = []
for item in list_a:
    if item in list_b:
        common.append(item)

for item in common:
    result.append(item)

print("Ergebnis:", result)

````

### Aufgabe 56. Sortieren ohne `.sort()` ⭐⭐⭐

Gegeben ist eine Zahlenliste. Sortiere sie aufsteigend, **ohne die Methode `.sort()` und die Funktion `sorted()` zu verwenden**.

> 💡 **Tipp:** Der einfachste Algorithmus ist „Bubble Sort": in einer verschachtelten Schleife vergleichen wir benachbarte Elemente und tauschen sie, wenn links das größere steht. Wir wiederholen, bis wir die ganze Liste ohne einen einzigen Tausch durchlaufen haben.


````python
numbers = [42, 17, 89, 3, 56, 91, 24]
# Dein Code:
n = len(numbers)
for i in range(n):
    for j in range(n - 1):
        if numbers[j] > numbers[j + 1]:
            # tauschen
            temp = numbers[j]
            numbers[j] = numbers[j + 1]
            numbers[j + 1] = temp

print("Sortierte Liste:", numbers)
````

### Aufgabe 58. Cäsar-Chiffre ⭐⭐⭐

Frage den Benutzer nach einem Wort (auf Deutsch oder Englisch) und einer Zahl `shift`.
Verschlüssele das Wort: ersetze jeden Buchstaben durch den, der `shift` Positionen weiter im Alphabet steht.

**Beispiel:** Wort `abc`, `shift = 2` → `cde`.

> 💡 **Tipp:** Verwende die Funktionen `ord(letter)` (gibt den Zeichencode zurück) und `chr(num)` (gibt das Zeichen zum Code zurück). Für jeden Buchstaben: neuer Code = ord(letter) + shift. Der Einfachheit halber musst du den Übergang über das Alphabetende hinaus nicht behandeln.


````python
# Dein Code für Aufgabe 58:
word = input("Wort eingeben: ")
shift = int(input("Shift eingeben: "))

encrypted_word = ""

for letter in word:
    new_letter = chr(ord(letter) + shift)
    encrypted_word += new_letter

print("Verschlüsseltes Wort:", encrypted_word)

````

### Aufgabe 59. Spiel „Zahl erraten" ⭐⭐⭐

Erstelle eine Variable `secret = 42` (die gesuchte Zahl). Frage in einer `while`-Schleife den Benutzer nach einer Zahl:
- Wenn die eingegebene Zahl kleiner als die gesuchte ist — schreibe „Größer"
- Wenn größer — „Kleiner"
- Wenn erraten — schreibe „Herzlichen Glückwunsch!" und beende die Schleife
- Zähle die Anzahl der Versuche und gib sie am Ende aus

> 💡 **Tipp:** Verwende eine Flag-Variable `guessed = False` als Schleifenbedingung. Wenn der Benutzer richtig geraten hat — `guessed = True`.


````python
# Dein Code für Aufgabe 59:
secret = 42
guessed = False
attempts = 0
while guessed == False:
    number = int(input("Zahl eingeben: "))
    attempts += 1
    if number < secret:
        print("Größer")
    elif number > secret:
        print("Kleiner")
    else:
        print("Herzlichen Glückwunsch!")
        guessed = True

print("Anzahl der Versuche:", attempts)
````

### Aufgabe 60. Textanalyse ⭐⭐⭐

1. Anzahl der Buchstaben (ohne Leerzeichen)
2. Anzahl der Wörter

> 💡 **Tipp:** Die Methode `.split()` zerlegt einen String in eine Liste von Wörtern. Zum Beispiel gibt `'hello world'.split()` `['hello', 'world']` zurück.


````python
text = "programmieren mit python ist interessant und nützlich"
# Dein Code:

letters_count = len(text.replace(" ", ""))
words = text.split()
words_count = len(words)
vowels_count = 0
for letter in text.lower():
    if letter in "aeiouäöü":
        vowels_count += 1
longest_word = ""
for word in words:
    if len(word) > len(longest_word):
        longest_word = word

print("Anzahl der Buchstaben:", letters_count)
print("Anzahl der Wörter:", words_count)
print("Anzahl der Vokale:", vowels_count)
print("Längstes Wort:", longest_word)

````

### Aufgabe 61. Geschäft: Verkaufsstatistik ⭐⭐⭐

Gegeben ist eine Liste mit Verkäufen für eine Woche (eine Zahl pro Tag, beginnend mit Montag).
Gib aus:
1. Den Gesamtumsatz für die Woche
2. Den durchschnittlichen Tagesumsatz
3. Den besten Tag (mit dem höchsten Umsatz) — mit Namen (Montag, Dienstag...)
4. Die Anzahl der Tage, an denen der Umsatz über dem Durchschnitt lag

> 💡 **Tipp:** Erstelle eine Liste mit den Tagesnamen: `days = ["Montag", "Dienstag", ...]`. Finde den Index des besten Tages in einer Schleife (oder mit `sales.index(max(sales))`).


````python
sales = [3200, 2800, 4100, 3500, 5200, 7800, 6500]
# Dein Code:
days = [
    "Montag",
    "Dienstag",
    "Mittwoch",
    "Donnerstag",
    "Freitag",
    "Samstag",
    "Sonntag"
]
total = sum(sales)
average = total / len(sales)
max_sale = max(sales)
best_day_index = sales.index(max_sale)
best_day = days[best_day_index]
count = 0

for sale in sales:
    if sale > average:
        count += 1

# Ausgabe der Ergebnisse
print("Gesamtumsatz für die Woche:", total)
print("Durchschnittlicher Tagesumsatz:", average)
print("Bester Tag:", best_day)
print("Anzahl der Tage über dem Durchschnittsumsatz:", count)
````

### Aufgabe 62. Tabellenstand ⭐⭐⭐

Gegeben sind zwei Listen: Teams und ihre Punkte (entsprechende Positionen).
Gib die Tabelle SORTIERT nach absteigenden Punkten in folgendem Format aus:
```
1. Bayern — 67 Punkte
2. Dortmund — 64 Punkte
...
```

> 💡 **Tipp:** Der einfachste Weg: finde in einer Schleife das Team mit den meisten Punkten, gib es aus, entferne dann dieses Paar (Team und Punkte) aus beiden Listen. Wiederhole, bis die Listen leer sind.


````python
teams = ["Dortmund", "Bayern", "Leipzig", "Leverkusen", "Frankfurt", "Freiburg"]
points = [64, 67, 52, 48, 39, 45]
counter = 0
# Dein Code:
while len(teams) > 0:
    indx_max = points.index(max(points))
    counter += 1
    print(counter, ". ", teams[indx_max]," - ",max(points))
    points.pop(indx_max)
    teams.pop(indx_max)

````
