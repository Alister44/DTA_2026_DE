### ✏️ Aufgabe 1.1


### ✏️ Aufgabe 2.1

Es gibt ein Array mit den täglichen Verkäufen eines Geschäfts über 2 Wochen (14 Tage):

```python
sales = np.array([320, 415, 290, 480, 350, 620, 580,
                  310, 440, 305, 510, 370, 650, 595])
```

Führe folgende Auswahlen **ohne Schleifen** aus, nur mit Slicing:
1. Verkäufe der ersten Woche (Tage 1–7)
2. Verkäufe der zweiten Woche (Tage 8–14)
3. Verkäufe am Wochenende: 6., 7., 13., 14. Tag — verwende die Indizierung mit einer Liste `[[5,6,12,13]]`

````python
sales = np.array([320, 415, 290, 480, 350, 620, 580,
                  310, 440, 305, 510, 370, 650, 595])

# Dein Code:
print(f"Verkäufe der ersten Woche (Tage 1–7) {sales[:7]}")
print(f"Verkäufe der zweiten Woche (Tage 8–14) {sales[7:]}")
print(f"Verkäufe am Wochenende: 6., 7., 13., 14. Tag {sales[[5,6,12,13]]}")
print(f"Verkäufe jeden zweiten Tag, beginnend mit dem ersten (ungerade Tage: 1, 3, 5...) {sales[0::2]}")
print(f"Array in umgekehrter Reihenfolge {sales[::-1]}")
````

> **Was wir wiederholen:** Arithmetik mit Arrays, `np.sqrt()`, `np.round()`,

### ✏️ Aufgabe 3.1

Es gibt ein Array mit Produktpreisen in Dollar. Führe **ohne Schleifen** aus:

```python
prices_usd = np.array([15.99, 42.50, 8.00, 125.00, 33.75, 67.20, 9.99])
```

1. Rechne die Preise in Euro um (Kurs: **1 USD = 0.92 EUR**), runde das Ergebnis auf 2 Stellen
2. Berechne 19% MwSt. — multipliziere die Euro-Preise mit 1.19, runde auf 2 Stellen
3. Finde: den minimalen, maximalen und durchschnittlichen Preis mit MwSt.
4. Finde die Summe aller Produkte mit MwSt. — das ist der Gesamtwert des Warenkorbs
5. Gib die Ergebnisse mit Erklärungen aus


````python
prices_usd = np.array([15.99, 42.50, 8.00, 125.00, 33.75, 67.20, 9.99])

# Dein Code:
prices_eur = (prices_usd * 0.92).round(2)
print(f"Preise in Euro (Kurs: 1 USD = 0.92 EUR) {prices_eur}")

prices_vat = (prices_eur * 1.19).round(2)
print(f"Preise mit 19% MwSt. {prices_vat}")

print(f"minimal {prices_vat.min()}, maximal {prices_vat.max()}, durchschnittlicher Preis {prices_vat.mean().round(2)} mit MwSt.")

print(f"Summe aller Produkte mit MwSt. {prices_vat.sum()}")

````

### ✏️ Aufgabe 3.2

Es gibt Daten zur Anzahl der Produkte im Lager und ihren Selbstkosten.
Führe **ohne Schleifen** aus:

```python
quantity  = np.array([120, 45, 300, 18, 75, 200, 60])
cost_each = np.array([250, 1800, 45, 9500, 380, 120, 750])
names     = ['Kabel', 'Tastatur', 'Notizbuch', 'Monitor',
             'Maus', 'Stift', 'Kopfhörer']
```

1. Berechne den Gesamtwert jedes Produkts im Lager (`quantity * cost_each`)
2. Finde den Gesamtwert des gesamten Lagers
3. Finde das Produkt mit dem höchsten Gesamtwert — verwende `np.argmax()`
4. Finde das Produkt mit dem niedrigsten Gesamtwert — `np.argmin()`
5. Gib den durchschnittlichen Wert einer Produktart im Lager aus


````python
quantity  = np.array([120, 45, 300, 18, 75, 200, 60])
cost_each = np.array([250, 1800, 45, 9500, 380, 120, 750])
names     = ['Kabel', 'Tastatur', 'Notizbuch', 'Monitor',
             'Maus', 'Stift', 'Kopfhörer']

# Dein Code:
total_cost = quantity * cost_each
print(f"Gesamtwert jedes Produkts im Lager: ", total_cost)

print(f"Gesamtwert des gesamten Lagers: ", total_cost.sum())
#print(f"Gesamtwert des gesamten Lagers: ", np.dot(quantity,cost_each))

print(f"Produkt mit dem höchsten Gesamtwert: ", names[np.argmax(total_cost)])
print(f"Produkt mit dem niedrigsten Gesamtwert: ", names[np.argmin(total_cost)])

print(f"durchschnittlicher Wert einer Produktart: ", total_cost.mean().round(2))
````

## Kapitel 4 — Boolesches Maskieren und Filterung

> **Was wir wiederholen:** boolesche Masken, Filterung `arr[mask]`,
> `np.where()`, `.sum()` zum Zählen von True


### ✏️ Aufgabe 4.1

Es gibt ein Array mit Bearbeitungszeiten von Bestellungen in Minuten (Norm — bis zu 30 Min):

```python
delivery_time = np.array([22, 35, 28, 41, 19, 55, 27, 33, 24, 48,
                           31, 26, 52, 18, 29, 38, 23, 44, 20, 36])
```

1. Finde alle Bestellungen, die **rechtzeitig** bearbeitet wurden (≤ 30 Min)
2. Finde alle **verspäteten** Bestellungen (> 30 Min)
3. Berechne die Anzahl der rechtzeitigen und der verspäteten
4. Finde den Prozentsatz der verspäteten Bestellungen (auf 1 Stelle runden)
5. Wie hoch ist die durchschnittliche Verspätung bei den verspäteten Bestellungen?


````python
delivery_time = np.array([22, 35, 28, 41, 19, 55, 27, 33, 24, 48,
                           31, 26, 52, 18, 29, 38, 23, 44, 20, 36])

# Dein Code:
in_time = delivery_time <= 30
print("rechtzeitig bearbeitete Bestellungen (≤ 30 Min): ",delivery_time[in_time])

not_in_time = delivery_time > 30
print("alle verspäteten Bestellungen (> 30 Min): ",delivery_time[not_in_time])

print(f"Anzahl rechtzeitig {in_time.sum()} und verspätet {not_in_time.sum()}")
print(f"Prozentsatz der verspäteten Bestellungen {not_in_time.sum()/len(delivery_time) * 100:.1f}%")
print(f"durchschnittliche Verspätung bei verspäteten Bestellungen: {(delivery_time[not_in_time] - 30).mean()}")
````

### ✏️ Aufgabe 4.2

Es gibt ein Array mit täglichen Verkäufen für einen Monat. Verwende `np.where()`:

```python
np.random.seed(42)
daily = np.random.randint(100, 900, size=30)
```

1. Finde Tage mit Verkäufen **über dem Durchschnitt** für den Monat
2. Berechne, wie viele solcher Tage es gibt
3. Erstelle mit `np.where()` ein Array `status`:
   - `'Guter Tag'` — wenn die Verkäufe über dem Durchschnitt liegen
   - `'Normaler Tag'` — wenn darunter oder gleich
4. Gib die ersten 10 Tage aus: Verkäufe und ihr Status nebeneinander


````python
np.random.seed(42)
daily = np.random.randint(100, 900, size=30)

# Dein Code:
print(f"Tage mit Verkäufen über dem Durchschnitt: {daily[daily > daily.mean()]}")
print(f"solcher Tage: {(daily > daily.mean()).sum()}")
status = np.where(daily > daily.mean(),"Guter Tag","Normaler Tag")
print(status)
for sale, st in zip(daily[:10], status[:10]):
    print(f"{sale}  —  {st}")
````

## Kapitel 5 — reshape und Sortieren

> **Was wir wiederholen:** `.reshape()`, `.flatten()`, `np.sort()`, `np.argsort()`


### ✏️ Aufgabe 5.1

Es gibt ein 1D-Array mit 24 Verkaufswerten (nach Stunden über den Tag):

```python
hourly = np.array([12, 8, 5, 3, 4, 9, 18, 35, 52, 61, 58, 64,
                   70, 67, 55, 48, 53, 62, 74, 68, 45, 30, 22, 15])
```

1. Wandle das Array in eine Matrix **6×4** um (6 Blöcke à 4 Stunden) und gib sie aus
2. Finde die Summe der Verkäufe in jedem 4-Stunden-Block (pro Zeile)
3. Sortiere das ursprüngliche 1D-Array **aufsteigend** und gib es aus
4. Finde die **Indizes** der Stunden, sortiert nach aufsteigenden Verkäufen — `np.argsort()`

## Kapitel 6 — Erzeugung von Zufallszahlen

> **Was wir wiederholen:** `np.random.seed()`, `np.random.randint()`,
> `np.random.normal()`, `np.random.choice()`


### ✏️ Aufgabe 6.1

Simuliere die Ergebnisse einer Marketingumfrage unter 50 Kunden.

1. Setze `np.random.seed(99)`
2. Generiere das Array `ages` — Alter von 50 Kunden: ganze Zahlen von 18 bis 65
3. Generiere das Array `scores` — Zufriedenheitsbewertungen (Normalverteilung, Mittelwert=7, σ=1.5), auf 1 Stelle runden
4. Begrenze `scores` so, dass die Werte zwischen 1 und 10 liegen — `np.clip(arr, 1, 10)`
5. Gib aus:
   - das durchschnittliche Alter der Kunden
   - die durchschnittliche Zufriedenheitsbewertung
   - die Anzahl der Kunden mit einer Bewertung >= 8 („zufrieden")
   - den Prozentsatz der Zufriedenen


````python
# Dein Code:
rng = np.random.default_rng(seed=99)
ages = rng.integers(18, 66, size=50)
scores = rng.normal(loc=7, scale=1.5, size=50).round(1)
# np.clip begrenzt die Werte des Arrays auf einen vorgegebenen Bereich
#  — alles unter dem Minimum wird zum Minimum, alles über dem Maximum wird zum Maximum.
scores_new = np.clip(scores, 1, 10)
print(f"durchschnittliches Alter der Kunden: {ages.mean().round()}")
print(f"durchschnittliche Zufriedenheitsbewertung: {scores_new.mean().round(2)}")
print(f"Anzahl der Kunden mit einer Bewertung >= 8 („zufrieden\"): {(scores_new >= 8).sum()}")
print(f"Prozentsatz der Zufriedenen: {(scores_new >= 8).sum() / len(scores_new) * 100} %")
````

## 🏆 Abschlussprojekt — Analyse der Shop-Daten

> **Wir verwenden alles aus dem Tutorial zusammen**

Du bist Datenanalyst eines kleinen Elektronikgeschäfts.
Du hast ein Array mit Verkäufen über 8 Wochen (56 Tage) erhalten.

Führe eine vollständige Analyse durch und beantworte die Fragen deines Vorgesetzten.


````python
# Daten — ändere diese Zelle nicht
np.random.seed(2024)
base = np.random.randint(200, 800, 56)

# Am Wochenende verkauft sich besser
weekend_boost = np.array([1 if i % 7 in [5, 6] else 0 for i in range(56)])
sales_56 = (base + weekend_boost * np.random.randint(100, 300, 56)).astype(int)

print('Daten erhalten. Array sales_56 ist bereit für die Analyse.')
print(f'Elemente: {sales_56.size}, Form: {sales_56.shape}')

````

### Frage 1 — Allgemeine Statistik (Kapitel 1, 3)
Gib aus:
- Gesamtumsatz über 56 Tage
- Durchschnittlicher Tagesumsatz (auf 2 Stellen runden)
- Minimaler und maximaler Tagesverkauf
- Standardabweichung (wie stark die Verkäufe „schwanken")


````python
# Dein Code:
print("Gesamtumsatz über 56 Tage",np.sum(sales_56))
print("Durchschnittlicher Tagesumsatz (auf 2 Stellen gerundet)",np.mean(sales_56).round(2))
print(f"Minimaler Tagesverkauf - {np.min(sales_56)} und maximaler Tagesverkauf {np.max(sales_56)}")
print("Standardabweichung (wie stark die Verkäufe „schwanken\")",np.std(sales_56).round(2))
````

### Frage 2 — Wochenanalyse (Kapitel 5)
1. Wandle `sales_56` in eine Matrix **8×7** um (8 Wochen à 7 Tage)
2. Finde den Umsatz jeder Woche (Summe pro Zeile)
3. Finde die beste Woche und gib ihre Nummer (gezählt ab 1) und die Summe aus
4. Finde den durchschnittlichen Tagesumsatz für jeden Wochentag (Durchschnitt pro Spalte)
5. Gib aus, welcher Wochentag im Durchschnitt am profitabelsten ist (0=Mo, 6=So)


````python
# Dein Code:
matx8x7 = sales_56.reshape(8,7)
print("Matrix nach Wochen ",matx8x7)
print("Umsatz jeder Woche ", matx8x7.sum(axis=1))
print(f"beste Woche - {np.argmax(matx8x7.sum(axis=1))+1}, ihre Nummer (gezählt ab 1) und Summe {np.max(matx8x7.sum(axis=1))}")
print("durchschnittlicher Tagesumsatz für jeden Wochentag (Durchschnitt pro Spalte) - ", matx8x7.sum(axis=0))
print("Wochentag, der im Durchschnitt am profitabelsten ist (0=Mo, 6=So) ", np.argmax(matx8x7.sum(axis=0)))
````

### Frage 3 — Filterung und Maskierung (Kapitel 4)
1. Finde die Anzahl der **„schwachen" Tage** — Verkäufe unter 400 €
2. Finde die Anzahl der **„starken" Tage** — Verkäufe über 700 €
3. Wie hoch ist der durchschnittliche Umsatz an schwachen und wie hoch an starken Tagen?
4. Erstelle ein Array `labels` mit `np.where()`:
   - `'Schwach'` wenn < 400
   - `'Stark'` wenn > 700
   - `'Normal'` in allen anderen Fällen (Tipp: verschachteltes `np.where`)
5. Gib die ersten 14 Tage aus: Verkäufe und ihre Bezeichnungen


````python
# Dein Code:
print("Anzahl der „schwachen\" Tage — Verkäufe unter 400 €: ", (sales_56 < 400).sum())
print("Anzahl der „starken\" Tage — Verkäufe über 700 €: ", (sales_56 > 700).sum())
print(f"durchschnittlicher Umsatz an schwachen Tagen - {sales_56[sales_56 < 400].mean().round(2)} an starken - {sales_56[sales_56 > 700].mean().round(2)}" )
labels = np.where(sales_56 < 400, "Schwach", np.where(sales_56 > 700, "Stark", "Normal"))
print(f"erste 14 Tage: Verkäufe und ihre Bezeichnungen ")
for sale, label in zip(sales_56[:14], labels[:14]):
    print(f"{sale} - {label}")
````
