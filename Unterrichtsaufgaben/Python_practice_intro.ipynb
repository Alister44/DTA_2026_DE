Die Funktion `range` in Kombination mit der `for`-Schleife ist in Python ein mächtiges Werkzeug, um die Wiederholung von Aktionen eine bestimmte Anzahl von Malen zu steuern. Dieser Mechanismus wird häufig zum Iterieren über eine Zahlenfolge verwendet, was ihn besonders nützlich für verschiedene Aufgaben macht, von einfachen bis zu komplexen Algorithmen.



Die Funktion range erzeugt eine Zahlenfolge. Sie kann auf unterschiedliche Weise verwendet werden:



`range(stop)`: Erzeugt eine Zahlenfolge von 0 bis stop - 1.

`range(start, stop)`: Erzeugt Zahlen von start bis stop - 1.

`range(start, stop, step)`: Erzeugt Zahlen von start bis stop - 1, mit der Schrittweite step.


die `for`-Schleife zusammen mit `range` zum Iterieren über eine Zahlenfolge sieht so aus:


````python
range (0,5)
````

````python
for el in range(5):
  print(el)
````

````python
for el in range(2,5):
  print(el)
````

````python
for el in range(2,15,3):
  print(el)
````

````python
for el in range(5,2,-1):
  print(el)
````

````python
for el in range(3,0,-2):
  print(el)
````

````python
num = int(input("Enter number "))
for i in range(1, num + 1):
  print(i, end = " / ")
````

````python
print(1,2,3,4,5, sep = ' / ')
````

````python
print(*range(1, num + 1), sep = ' / ')
````

# Kapitel 1. Wörterbücher (`dict`)

**Kurze Wiederholung:**

Ein Wörterbuch (dict) ist eine Sammlung von `Schlüssel : Wert`-Paaren, erstellt in geschweiften Klammern:
```python
student = {"name": "Olena", "age": 20, "city": "Kyjiw"}
```

**Grundlegende Operationen:**
- Zugriff: `student["name"]`
- Hinzufügen/Ändern: `student["email"] = "olena@example.com"`
- Löschen: `del student["age"]`
- Prüfung: `if "name" in student:`
- Sicherer Zugriff: `student.get("phone")` — gibt `None` zurück, wenn der Schlüssel nicht existiert
- Länge: `len(student)` — Anzahl der Paare

**Durchlaufen in einer Schleife:**
- `for key in d.keys():` — Durchlaufen der Schlüssel
- `for value in d.values():` — Durchlaufen der Werte
- `for key, value in d.items():` — Durchlaufen der Paare


`dict()` - Wörterbuch  
`d = {}` - leeres Wörterbuch erstellt  
`d = {'key': 'value', 'key2': 'value2'}`
`key` - unveränderliche Daten  
In einem Wörterbuch dürfen keine 2 gleichen Schlüssel vorkommen!!  
`value` - kann geändert werden oder auch leer gelassen werden.
