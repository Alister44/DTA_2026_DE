arr[arr > 5]             # nur Elemente > 5
arr[arr < 0] = 0         # negative durch Null ersetzen
mask = arr > 5
arr[mask]                # Maske anwenden
arr[~mask]               # invertierte Maske (NICHT)

# Mehrere Bedingungen:
arr[(arr > 5) & (arr < 20)]   # UND
arr[(arr < 0) | (arr > 100)]  # ODER

# ── Nützliche Techniken ────────────────────────────────────
mask.sum()               # Anzahl True (wie viele erfüllen die Bedingung)
mask.mean() * 100        # Prozentsatz True

arr.copy()               # Kopie (Änderungen wirken sich nicht auf das Original aus)

# Min-Max-Normalisierung:
(arr - arr.min()) / (arr.max() - arr.min())

# Ausreißer-Erkennung (2σ-Regel):
np.abs(arr - arr.mean()) > 2 * arr.std()

# Z-Score (Standardisierung):
(arr - arr.mean()) / arr.std()
