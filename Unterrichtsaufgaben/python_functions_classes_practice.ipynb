### Aufgabe 10. Primzahlprüfung ⭐⭐

Erstelle eine Funktion `is_prime(n)`, die `True` zurückgibt, wenn die Zahl eine Primzahl ist, und sonst `False`.

**Erinnerung:** eine Primzahl ist nur durch 1 und sich selbst teilbar. Zahlen kleiner als 2 sind keine Primzahlen.

**Prüfe:**
```python
print(is_prime(7))    # True
print(is_prime(10))   # False
print(is_prime(1))    # False
```

> 💡 **Tipp:** Prüfe in einer Schleife von 2 bis n-1, ob `n` ohne Rest durch eine Zahl teilbar ist. Wenn du einen Teiler findest — gib sofort `False` zurück. Wenn die Schleife ohne gefundenen Teiler endet — `True`.


````python
# Dein Code für Aufgabe 10:
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True
print(is_prime(7))
print(is_prime(10))
print(is_prime(1))

````

### Aufgabe 13. Arithmetischer Mittelwert ⭐⭐

Erstelle eine Funktion `average(numbers)`, die eine Zahlenliste entgegennimmt und ihren arithmetischen Mittelwert zurückgibt. Wenn die Liste leer ist — gib 0 zurück.

> 💡 **Tipp:** Vergiss nicht, die leere Liste vor der Division zu prüfen, um eine Division durch Null zu vermeiden.


````python
# Dein Code für Aufgabe 13:

def average(numbers):
    if numbers:
        sum_n = 0
        for el in numbers:
            sum_n += el
        return sum_n / len(numbers)
    else:
        return 0
print(average([3,4,6]))
print(average([]))
print(average([0]))
````

Erstelle eine Funktion `count_digits(n)`, die eine ganze Zahl entgegennimmt und die Anzahl ihrer Ziffern zurückgibt.

**Prüfe:**
```python

print(count_digits(7))       # 1
print(count_digits(0))       # 1
```

Implementiere es über **Mathematik** (ohne Umwandlung in einen String).

> 💡 **Tipp:** Teile in der Schleife `while n > 0` die Zahl ganzzahlig durch 10 (`n = n // 10`) und erhöhe den Zähler. Behandle den Fall `n == 0` gesondert.


````python
# Dein Code für Aufgabe 14:
def count_digits(n):
    count = 0
    while n > 0:
        n = n // 10
        count += 1
    return count
count_digits(35233)
````

### Aufgabe 16. Längstes Wort ⭐⭐⭐

Erstelle eine Funktion `longest_word(sentence)`, die einen Satz-String entgegennimmt und das längste Wort zurückgibt.

**Prüfe:**
```python
print(longest_word("Python ist eine wunderbare Programmiersprache"))   # "Programmiersprache"
```

> 💡 **Tipp:** Zerlege den Satz mit `.split()` in eine Liste von Wörtern. Durchlaufe sie dann und finde das längste, indem du mit `len()` vergleichst.


````python
# Dein Code für Aufgabe 16:
def longest_word(sentence):
    words = sentence.split()
    max_w =  words[0]
    for word in words:
       if len(word) >len(max_w):
            max_w = word
    return max_w
print(longest_word("Python ist eine wunderbare Programmiersprache"))
````

### Aufgabe 17. Rechner mit Operation ⭐⭐⭐

Erstelle eine Funktion `calculate(a, b, operation)`, die je nach String `operation` eine der Operationen ausführt:
- `"+"` — Addition
- `"-"` — Subtraktion
- `"*"` — Multiplikation
- `"/"` — Division

Wenn die Operation unbekannt ist — gib den String `"Unbekannte Operation"` zurück. Bei Division durch Null — gib `"Fehler: Division durch Null"` zurück.


````python
# Dein Code für Aufgabe 17:
def calculate(a, b, operation):
    if operation == "+":
        return a + b
    elif operation == "-":
        return a - b
    elif operation == "*":
        return a * b
    elif operation == "/":
        if b == 0:
           return "Fehler: Division durch Null"
        else:
            return a / b
    else:
        return "Unbekannte Operation"
print(calculate(4,5,"+"))
print(calculate(4,5,"-"))
print(calculate(4,5,"*"))
print(calculate(4,5,"/"))
print(calculate(4,0,"/"))
print(calculate(4,0,"#"))
````

### Aufgabe 18. Fibonacci-Zahlen ⭐⭐⭐

Erstelle eine Funktion `fibonacci(n)`, die eine **Liste** der ersten `n` Fibonacci-Zahlen zurückgibt.

**Erinnerung:** jede Zahl ist die Summe der beiden vorherigen. Die ersten beiden sind: 0 und 1.

**Prüfe:**
```python
print(fibonacci(7))   # [0, 1, 1, 2, 3, 5, 8]
```

> 💡 **Tipp:** Erstelle eine Liste mit den ersten beiden Zahlen `[0, 1]`. Füge in einer Schleife die Summe der letzten beiden Elemente mit `.append()` hinzu.


````python
# Dein Code für Aufgabe 18:
def fibonacci(n):
    l_fib = [0, 1]
    for i in range(2,n):
        l_fib.append(l_fib[i-2] + l_fib[i-1])
    return l_fib
fibonacci(7)
````

````python
# Dein Code für Aufgabe 23:
class Student:
    def __init__ (self,name):
        self.name = name
        self.grades = []

    def add_grade(self,grade):
        self.grades.append(grade)

    def average(self):
        if not self.grades:
            return 0
        else:
             return sum(self.grades) / len(self.grades)
    def __str__ (self):
        return f"Student {self.name}, Durchschnitt: {self.average()}"


stud1 = Student("Peter")
stud1.add_grade(87)
stud1.add_grade(90)
stud1.add_grade(70)
stud1.add_grade(73)
print(stud1)
````

````python
# Dein Code für Aufgabe 25:
class BankAccount:
    def __init__ (self,owner):
        self.owner = owner
        self.balance = 0

    def deposit(self,amount):
        self.balance += amount

    def withdraw(self,amount):
        if self.balance >= amount:
            self.balance -= amount
        else:
            print("Nicht genug Geld auf dem Konto")

    def show_balance(self):
        print(f"Kontostand {self.owner}: {self.balance} €")

acc1 = BankAccount("Peter")
acc1.deposit(1000)
acc1.show_balance()
acc1.withdraw(1100)
acc1.show_balance()
acc1.deposit(1000)
acc1.withdraw(1100)
acc1.show_balance()
````

### Aufgabe 26. Klasse Book mit Zähler ⭐⭐

Erstelle eine Klasse `Book` mit den Attributen `title`, `author`, `pages`. Füge hinzu:
1. Das Attribut `current_page = 0` (aktuelle Seite)
2. Die Methode `read(p)` — fügt `p` Seiten zu `current_page` hinzu, aber nicht mehr als `pages` (Gesamtzahl)
3. Die Methode `progress()` — gibt den Prozentsatz des Gelesenen zurück (zum Beispiel `"45.5%"`)
4. `__str__()` — `"<title> von <author>: <current_page>/<pages>"`

> 💡 **Tipp:** Um `current_page` zu begrenzen, ist `min(self.current_page + p, self.pages)` praktisch. Für die Formatierung — f-string mit `{value:.1f}%`.


````python
# Dein Code für Aufgabe 26:
class Book:
    def __init__(self,title,author,pages):
        self.title =  title
        self.author =  author
        self.pages =  pages
        self.current_page = 0

    def read(self,p):
        if p > self.pages:
            print(f"Du kannst kein Lesezeichen auf Seite {p} setzen, das Buch hat nur {self.pages} Seiten")
        else:
            self.current_page = p

    def progress(self):
         prog = self.current_page / self.pages *100
         print(f"Progress in book - {prog:.1f}%")

    def __str__(self):
        return f"{self.title} von {self.author}: {self.current_page}/{self.pages} "

book1 = Book("Harry Potter","Joahn Roaling",450)
print(book1)
book1.read(456)
book1.read(445)
print(book1)
book1.progress()


````
