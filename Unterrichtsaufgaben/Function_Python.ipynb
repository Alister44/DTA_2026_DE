````python
my_dict = {
    1:'one',
    2:'two',
    3:'three',
    4:'four',
    5:'five'
}
````

````python
#Повертає NoneType,змінює початкові дані, виводить результат
def info(d):
  print(d)

#info(my_dict)
type(info(my_dict))
````

````python
def add_el_view(d):
  d[6] = "six"
  print(d)

#info(my_dict)
type(add_el_view(my_dict))
````

````python
def add_el(d):
  d[6] = "six"


print(type(add_el(my_dict)))
print(f"add_el:{my_dict}")
````
